```python
import numpy as np
import matplotlib.pyplot as plt

c = 3e8
f = 900e6
Pt = 1.0
Gt = Gr = 1.0
lam = c / f
k = 2 * np.pi / lam
d = np.logspace(1, np.log10(5000), 2000)


def dbm(p):
    return 10 * np.log10(p) + 30


def free_space(d):
    return Pt * Gt * Gr * (lam / (4 * np.pi * d)) ** 2


def two_ray_asymptotic(d, ht, hr):
    return Pt * Gt * Gr * (ht ** 2 * hr ** 2) / d ** 4


def two_ray_full(d, ht, hr, gamma=-1.0):
    d_los = np.sqrt(d ** 2 + (ht - hr) ** 2)
    d_ref = np.sqrt(d ** 2 + (ht + hr) ** 2)
    e_total = np.exp(-1j * k * d_los) / d_los + gamma * np.exp(-1j * k * d_ref) / d_ref
    return Pt * Gt * Gr * (lam / (4 * np.pi)) ** 2 * np.abs(e_total) ** 2


def breakpoint(ht, hr):
    return 4 * np.pi * ht * hr / lam


print(f"Wavelength = {lam:.4f} m")

ht, hr = 20.0, 2.0
d_break = breakpoint(ht, hr)
print(f"Breakpoint distance (ht={ht:.0f} m, hr={hr:.0f} m) = {d_break:.1f} m")

plt.figure()
plt.semilogx(d, dbm(free_space(d)), label="Free-space")
plt.semilogx(d, dbm(two_ray_asymptotic(d, ht, hr)), label="Two-ray asymptotic")
plt.semilogx(d, dbm(two_ray_full(d, ht, hr)), label="Two-ray full model", alpha=0.7)
plt.axvline(d_break, color="k", linestyle="--", label=f"Breakpoint = {d_break:.0f} m")
plt.xlabel("Distance (m)")
plt.ylabel("Received Power (dBm)")
plt.title(f"Free-space vs Two-ray (ht={ht:.0f} m, hr={hr:.0f} m)")
plt.ylim(-200, 0)
plt.grid(True, which="both")
plt.legend()
plt.savefig("two_ray_comparison.png", dpi=150, bbox_inches="tight")

ht2, hr2 = 50.0, 5.0
d_break2 = breakpoint(ht2, hr2)
print(f"Breakpoint distance (ht={ht2:.0f} m, hr={hr2:.0f} m) = {d_break2:.1f} m")

plt.figure()
plt.semilogx(d, dbm(free_space(d)), label="Free-space")
plt.semilogx(d, dbm(two_ray_asymptotic(d, ht2, hr2)), label="Two-ray asymptotic")
plt.semilogx(d, dbm(two_ray_full(d, ht2, hr2)), label="Two-ray full model", alpha=0.7)
plt.axvline(d_break2, color="k", linestyle="--", label=f"Breakpoint = {d_break2:.0f} m")
plt.xlabel("Distance (m)")
plt.ylabel("Received Power (dBm)")
plt.title(f"Free-space vs Two-ray (ht={ht2:.0f} m, hr={hr2:.0f} m)")
plt.ylim(-200, 0)
plt.grid(True, which="both")
plt.legend()
plt.savefig("two_ray_comparison_alt_heights.png", dpi=150, bbox_inches="tight")

print("\nChallenge task: breakpoint distance for different transmitter heights (hr = 2 m)")
print(f"{'ht (m)':>7} | {'d_break (m)':>12}")
print("-" * 22)
heights = [10.0, 20.0, 50.0]
plt.figure()
for h in heights:
    db = breakpoint(h, hr)
    print(f"{h:>7.0f} | {db:>12.1f}")
    line, = plt.semilogx(d, dbm(two_ray_full(d, h, hr)), label=f"ht = {h:.0f} m")
    plt.axvline(db, color=line.get_color(), linestyle="--", alpha=0.6)
plt.semilogx(d, dbm(free_space(d)), "k:", label="Free-space")
plt.xlabel("Distance (m)")
plt.ylabel("Received Power (dBm)")
plt.title("Effect of Transmitter Height (hr = 2 m)")
plt.ylim(-200, 0)
plt.grid(True, which="both")
plt.legend()
plt.savefig("two_ray_tx_height_challenge.png", dpi=150, bbox_inches="tight")

plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Parameters
c = 3e8
f = 2.4e9
Pt = 1.0
Gt = 1.0
Gr = 1.0
ht = [10, 20, 50]       # Transmitter heights (m)
hr = 2                  # Receiver height (m)

# Distance range
d = np.linspace(10, 10000, 1000)

# Wavelength
lam = c / f

# Free-space received power
Pr_fs = Pt * Gt * Gr * (lam / (4 * np.pi * d))**2
Pr_fs_dBm = 10 * np.log10(Pr_fs) + 30

# Plot received power
plt.figure(figsize=(10, 6))

plt.semilogx(d, Pr_fs_dBm, '--', label="Free Space", linewidth=2)

# Calculate and plot two-ray model for different heights
breakpoints = {}

for h in ht:

    # Breakpoint distance
    d_bp = (4 * np.pi * h * hr) / lam
    breakpoints[h] = d_bp

    # Two-ray ground reflection received power
    Pr_two_ray = Pt * Gt * Gr * (h * hr)**2 / (d**4)
    Pr_two_ray_dBm = 10 * np.log10(Pr_two_ray) + 30

    plt.semilogx(
        d,
        Pr_two_ray_dBm,
        label=f"Two-Ray (Ht={h} m)",
        linewidth=2
    )

    plt.axvline(
        d_bp,
        linestyle=':',
        label=f"Breakpoint Ht={h} m ({d_bp:.1f} m)"
    )

    print(f"Transmitter Height: {h} m")
    print(f"Breakpoint Distance: {d_bp:.2f} m\n")

plt.xlabel("Distance (m)")
plt.ylabel("Received Power (dBm)")
plt.title("Free-Space and Two-Ray Propagation Model")
plt.legend()
plt.grid(True, which="both")
plt.tight_layout()
plt.show()

# Breakpoint comparison
plt.figure(figsize=(8, 5))

heights = list(breakpoints.keys())
distances = list(breakpoints.values())

plt.bar(
    [str(h) + " m" for h in heights],
    distances
)

plt.xlabel("Transmitter Height (m)")
plt.ylabel("Breakpoint Distance (m)")
plt.title("Breakpoint Distance vs Transmitter Height")
plt.grid(axis="y")
plt.tight_layout()
plt.show()

# Display comparison
print("BREAKPOINT DISTANCE COMPARISON")
print("--------------------------------")

for h, bp in breakpoints.items():
    print(f"Height = {h} m : Breakpoint = {bp:.2f} m")
```
